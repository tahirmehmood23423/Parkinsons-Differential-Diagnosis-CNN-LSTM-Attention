```python
count = df.groupby(['label', 'gender']).size().unstack()
count.plot(kind='bar')
plt.title("Bar chart of Age with Categories")
```

```python
count = df.groupby(['label', 'appearance_in_kinship']).size().unstack()
count.plot(kind='bar')
plt.title("Bar chart of appearance in kinship with Categories")
```
